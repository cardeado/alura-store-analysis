### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

```python
import matplotlib.pyplot as plt

# Dados de faturamento
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
faturamento = [1534509.12, 1488459.06, 1464025.03, 1384497.58]

# Gráfico de barras
plt.figure(figsize=(8, 6))
plt.bar(lojas, faturamento, color=['blue', 'green', 'red', 'orange'])
plt.title('Faturamento Total por Loja')
plt.xlabel('Lojas')
plt.ylabel('Faturamento (R$)')
plt.show()
```

# 2. Vendas por Categoria


```python
# Contabilizando as categorias mais populares para cada loja
categoria_loja = loja['Categoria do Produto'].value_counts()
categoria_loja2 = loja2['Categoria do Produto'].value_counts()
categoria_loja3 = loja3['Categoria do Produto'].value_counts()
categoria_loja4 = loja4['Categoria do Produto'].value_counts()

# Exibindo as categorias mais populares
print("Categorias mais populares - Loja 1:\n", categoria_loja)
print("Categorias mais populares - Loja 2:\n", categoria_loja2)
print("Categorias mais populares - Loja 3:\n", categoria_loja3)
print("Categorias mais populares - Loja 4:\n", categoria_loja4)
```

```python
import matplotlib.pyplot as plt

# Definir categorias e contagem de produtos para cada loja
categorias_loja = ['moveis', 'eletronicos', 'brinquedos', 'eletrodomesticos', 'esporte e lazer', 'instrumentos musicais', 'livros', 'utilidades domesticas']
contagem_loja1 = [465, 448, 324, 312, 284, 182, 173, 171]

categorias_loja2 = ['moveis', 'eletronicos', 'brinquedos', 'eletrodomesticos', 'esporte e lazer', 'instrumentos musicais', 'livros', 'utilidades domesticas']
contagem_loja2 = [442, 422, 313, 305, 275, 224, 197, 181]

categorias_loja3 = ['moveis', 'eletronicos', 'brinquedos', 'eletrodomesticos', 'esporte e lazer', 'instrumentos musicais', 'livros', 'utilidades domesticas']
contagem_loja3 = [499, 451, 315, 278, 277, 177, 185, 177]

categorias_loja4 = ['moveis', 'eletronicos', 'brinquedos', 'esporte e lazer', 'eletrodomesticos', 'utilidades domesticas', 'livros', 'instrumentos musicais']
contagem_loja4 = [480, 451, 338, 277, 254, 201, 187, 170]


# Função para criar gráfico de pizza
def plot_pizza(categorias, contagem, loja_nome):
    plt.figure(figsize=(7, 7))
    plt.pie(contagem, labels=categorias, autopct='%1.1f%%', startangle=90)
    plt.title(f'Distribuição de Categorias - {loja_nome}')
    plt.ylabel('')  # Remove o rótulo "Categoria do Produto"
    plt.show()

# Gerar gráficos de pizza para cada loja
plot_pizza(categorias_loja1, contagem_loja1, 'Loja 1')
plot_pizza(categorias_loja2, contagem_loja2, 'Loja 2')
plot_pizza(categorias_loja3, contagem_loja3, 'Loja 3')
plot_pizza(categorias_loja4, contagem_loja4, 'Loja 4')
```

Categorias Mais Populares por Loja

Com base nos gráficos de pizza gerados, podemos observar que as categorias "móveis" e "eletrônicos" dominam as vendas em todas as lojas, seguidas por "brinquedos" e "eletrodomésticos". No entanto, é possível notar pequenas variações:

Loja 3 tem o maior número de vendas na categoria "móveis", o que indica uma forte preferência dos clientes por esse segmento.

Loja 4, embora tenha um desempenho geral mais fraco no faturamento, ainda mantém um padrão de categorias semelhante às demais.

As categorias "livros", "instrumentos musicais" e "utilidades domésticas" aparecem consistentemente como as menos populares.

Essas observações ajudam a complementar a análise de desempenho geral das lojas, com base não apenas em valores financeiros, mas também nas preferências dos clientes.
